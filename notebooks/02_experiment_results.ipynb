# Deployment trade-off analysis
Only run after `edgeai benchmark`. Plot measured values; do not treat smoke-test data as research outcomes.

```python
import pandas as pd
from pathlib import Path
p = Path("../results/benchmark.csv")
assert p.is_file(), "Run actual model benchmark first"
df = pd.read_csv(p)
assert df.run_type.eq("real_labeled_evaluation").all()
df[["backend","device","batch_size","top1_accuracy","median_latency_ms","p95_latency_ms","model_file_mb"]]
```

```python
import matplotlib.pyplot as plt
cpu=df[(df.device=="cpu") & (df.batch_size==1)]
ax=cpu.plot.bar(x="backend",y="median_latency_ms",legend=False,figsize=(8,4))
ax.set_ylabel("Measured CPU latency (ms)")
plt.tight_layout()
```
