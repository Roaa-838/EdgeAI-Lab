# Dataset integrity and preprocessing
Run after downloading Imagenette. Training/calibration live in train; held-out evaluation lives in val.

```python
from pathlib import Path
from edgeai.dataset import classes_in_directory, list_images, stratified_limit, preprocess
root = Path("../data/imagenette2-160")
classes = classes_in_directory(root / "train")
print(classes)
```

```python
mapping={name:i for i,name in enumerate(classes)}
train=list_images(root/"train",mapping)
val=list_images(root/"val",mapping)
print(f"train={len(train)}, held-out val={len(val)}")
print("Preprocessed input:", preprocess(val[0]["path"]).shape)
```
